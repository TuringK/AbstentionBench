```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
from pathlib import Path

DATA_ROOT = Path("../data/abstention_classifier_datasets")
SPLITS     = ["train", "val", "test"]

LABEL_META = {
    "A": "Categorical Refusal",
    "B": "Epistemic Deflection",
    "C": "Insufficient Information",
    "D": "Context-Faithful Answer",
    "E": "Direct Factual Answer",
    "F": "Hallucination/Confabulation",
    "G": "Adversarial/Format",
    "H": "Composite",
}

sns.set_theme(style="whitegrid", palette="tab10")
```

```python
def load_variant(variant: str) -> dict[str, pd.DataFrame]:
    """Load train/val/test for a given variant ('eight_class' or 'seven_class')."""
    dfs = {}
    for split in SPLITS:
        path = DATA_ROOT / variant / f"{split}.csv"
        dfs[split] = pd.read_csv(path)
        print(f"[{variant}] {split:5s}: {len(dfs[split]):>7,} rows | cols: {list(dfs[split].columns)}")
    return dfs

print("Eight-class:")
eight = load_variant("eight_class")

print("\nSeven-class:")
seven = load_variant("seven_class")
```
